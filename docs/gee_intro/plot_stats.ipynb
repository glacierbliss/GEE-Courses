```python
#plot_stats experiments with plotting in Jupyter notebooks
#TODO: better way to handle paths

# Import required libraries
import pandas as pd
import matplotlib.pyplot as plt

# Load the CSV file
df = pd.read_csv(r'C:\Users\andyb\Documents\U\GEE-Courses\data\dem_stats.csv')

# Verify the data
print("First few rows of the dataset:")
print(df.head())
```

```python
# Create a scatter plot of area vs mean
plt.figure(figsize=(10, 6))
plt.scatter(df['AWATER'], df['mean'], color='blue', alpha=0.5)

# Add text labels with the 'name' column
for i, row in df.iterrows():
    plt.text(row['AWATER'], row['mean'], row['STUSPS'], fontsize=9, ha='right', va='bottom')

# Customize the plot
plt.xlabel('Area Water?')
plt.ylabel('Mean')
plt.title('Scatter Plot of Area Water vs Mean with Name Labels')
plt.grid(True)

# Show the plot
plt.show()
```
